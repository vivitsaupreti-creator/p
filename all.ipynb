```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv("Telco-Customer-Churn.csv")
df.head()
```

```python
df.info()
```

```python
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"], errors="coerce")
```

```python
df.dropna(inplace=True)
```

```python
X = df.drop(["customerID", "Churn"], axis=1)
Y = df["Churn"]
```

```python
X = pd.get_dummies(X, drop_first=True)
```

```python
X.columns
```

```python
pd.get_dummies(df, columns=["gender", "Partner"])
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, Y_train, Y_test = train_test_split(
 X,Y, test_size=0.25
)
```

```python
print(X_train.head())   
print(X_train.dtypes)
```

```python
len(X_train),len(X_test)
```

```python
print(X.columns)
```

```python
from sklearn.preprocessing import StandardScaler

sc = StandardScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)
```

```python
print(df.isnull().sum())
print(X.isnull().sum()) 
print(X.isnull().sum().sum())
```

```python
from sklearn.neighbors import KNeighborsClassifier
model= KNeighborsClassifier()
model.fit(X_train,Y_train)
Y_pred= model.predict(X_test)
```

```python
Y_pred
```

```python
from sklearn.metrics import accuracy_score
print(accuracy_score(Y_test,Y_pred)*100)
```

```python
from sklearn.tree import DecisionTreeClassifier
model_dt= DecisionTreeClassifier()
model_dt.fit(X_train,Y_train)
Y_pred_dt= model_dt.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score
print(accuracy_score(Y_test,Y_pred)*100)
```

```python
from sklearn.ensemble import RandomForestClassifier
model_rm= RandomForestClassifier()
model_rm.fit(X_train,Y_train)
Y_pred_rm= model_rm.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score
print(accuracy_score(Y_test,Y_pred)*100)
```

```python
from sklearn.metrics import classification_report
knn_report= classification_report(Y_test,Y_pred)
dt_report= classification_report(Y_test,Y_pred_dt)
rf_report= classification_report(Y_test,Y_pred_rm)
```

```python
print('knn report',knn_report)
```

```python
print('dt report', dt_report)
```

```python
print('rf report' ,rf_report)
```

```python
from sklearn.naive_bayes import BernoulliNB
model_nb= BernoulliNB()
model_nb.fit(X_train,Y_train)
Y_pred_nb= model_nb.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score
print(accuracy_score(Y_test,Y_pred_nb)*100)
```

```python
from sklearn.svm import SVC
model_svm= SVC()
model_svm.fit(X_train,Y_train)
Y_pred_svm= model_svm.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score
print(accuracy_score(Y_test,Y_pred_svm)*100)
```

```python
from sklearn.metrics import classification_report
nb_report= classification_report(Y_test,Y_pred_nb)
svm_report = classification_report(Y_test,Y_pred_svm)
```

```python
print('nb report', nb_report)
```
