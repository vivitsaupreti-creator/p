```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
df= pd.read_csv("Telco-Customer-Churn.csv")
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
df.TotalCharges= pd.to_numeric(df.TotalCharges, errors="coerce")
df.dropna(inplace=True)
```

```python
X= df.drop(["customerID","Churn"],axis=1)
Y= df["Churn"]
```

```python
X= pd.get_dummies(X, drop_first=True)
```

```python
X_train, X_test, Y_train, Y_test= train_test_split(X,Y,test_size=0.3, random_state=42)
```

```python
model_orig= LogisticRegression()
model_orig.fit(X_train,Y_train)
Y_model_orig= model_orig.predict(X_test)
print(accuracy_score(Y_test,Y_model_orig)*100)
```

```python
from sklearn.decomposition import PCA
pca = PCA(n_components=5)
X_train_pca= pca.fit_transform(X_train)
X_test_pca= pca.transform(X_test)
```

```python
explained_variance_ratio = pca.explained_variance_ratio_
print(explained_variance_ratio)
```

```python
model_pca= LogisticRegression()
model_pca.fit(X_train_pca,Y_train)
Y_pred_pca= model_pca.predict(X_test_pca)
print(accuracy_score(Y_test,Y_pred_pca)*100)
```

```python
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
lda = LinearDiscriminantAnalysis()
X_train_lda= lda.fit_transform(X_train, Y_train)
X_test_lda= lda.transform(X_test)
```

```python
explained_variance_ratio = lda.explained_variance_ratio_
```

```python
model_lda= LogisticRegression()
model_lda.fit(X_train_lda,Y_train)
Y_model_lda= model_lda.predict(X_test_lda)
print(accuracy_score(Y_test,Y_model_lda)*100)
```

                   APPLING KPCA     

```python
from sklearn.decomposition import KernelPCA
kpca= KernelPCA(n_components=10)
X_train_kpca= kpca.fit_transform(X_train)
X_test_kpca= kpca.transform(X_test)
```

```python
model_kpca= LogisticRegression()
model_kpca.fit(X_train_kpca,Y_train)
Y_pred_kpca= model_kpca.predict(X_test_kpca)
print(accuracy_score(Y_test,Y_pred_kpca)*100)
```

```python
                                APPLING QDA
```

```python
from sklearn.discriminant_analysis import QuadraticDiscriminantAnalysis
qda = QuadraticDiscriminantAnalysis(reg_param=0.1)
qda.fit(X_train, Y_train)
Y_pred_qda = qda.predict(X_test)
print(accuracy_score(Y_test, Y_pred_qda) * 100)
```
